```python
experiments = {
    "1" : "Today the weather is",
    "2" : "Write exactly three colors separated by commas:",
    "3" : "He saw the bat and"
}
```

```python
def generate_text(prompt,mode="greedy",max_new_tokens=50,temperature=0.8,top_k=50,seed=42):
    inputs = tokenizer(prompt, return_tensors="pt")
    generation_kwargs = {
        "max_new_tokens": max_new_tokens,
        "pad_token_id": tokenizer.eos_token_id
    }

    if mode == "greedy":
        generation_kwargs["do_sample"] = False

    elif mode == "sampling":
        generation_kwargs["do_sample"] = True
        generation_kwargs["temperature"] = temperature
        generation_kwargs["top_k"] = top_k

        if seed is not None:
            torch.manual_seed(seed)
            if torch.cuda.is_available():
                torch.cuda.manual_seed_all(seed)

    model.eval()

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            **generation_kwargs
        )

    full_ids = outputs[0].tolist()
    input_length = inputs["input_ids"].shape[1]
    new_ids = full_ids[input_length:]
    full_text = tokenizer.decode(
        full_ids,
        skip_special_tokens=True
    )

    continuation = tokenizer.decode(
        new_ids,
        skip_special_tokens=True
    )
    return {
        "prompt": prompt,
        "mode": mode,
        "full_ids": full_ids,
        "new_ids": new_ids,
        "full_text": full_text,
        "continuation": continuation,
        "input_token_count": input_length,
        "generated_token_count": len(new_ids)
    }
```

```python
for key,prompt in experiments.items():
  greedy_res = generate_text(prompt)
  sampling_res = generate_text(prompt,"sampling")

  print(greedy_res["full_text"])
  print(sampling_res["full_text"])
```

#ОБУЧЕНИЕ СВОЕЙ МОДЕЛИ НА ПОДОБИЕ GPT2


## 0. Параметры обучения

```python
import hashlib
import json
import math
import shutil
import subprocess
import sys
import unicodedata
import re

from collections import Counter
from pathlib import Path

import torch
import transformers
import datasets
import tokenizers

from datasets import Dataset, DatasetDict, load_dataset
from huggingface_hub import HfApi

from tokenizers import (
    Tokenizer,
    models,
    trainers,
    pre_tokenizers,
    decoders,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    PreTrainedTokenizerFast,
    GPT2Config,
    GPT2LMHeadModel,
    DataCollatorForLanguageModeling,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback,
    set_seed,
)

from tqdm.auto import tqdm


SEED = 42

DATASET_ID = "wikimedia/wikipedia"
DATASET_CONFIG = "20231101.ru"
DOCUMENT_COUNT = 5000

VOCAB_SIZE = 8000
BLOCK_SIZE = 256

EOS_TOKEN = "<|endoftext|>"
PAD_TOKEN = "<|pad|>"

EPOCHS = 10
MICROBATCH_SIZE = 16
GRADIENT_ACCUMULATION = 4
LEARNING_RATE = 3e-4

RUN_NAME = "experiment_001"

ROOT = Path("/content/gpt2_project")
DATA_DIR = ROOT / "data"
TOKENIZER_DIR = ROOT / "tokenizer"
RUN_DIR = ROOT / "runs" / RUN_NAME
FINAL_DIR = RUN_DIR / "final"

for directory in [DATA_DIR, TOKENIZER_DIR, RUN_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

set_seed(SEED)

print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Папка проекта:", ROOT)
```

## 1. Подготовка корпуса


```python
source = load_dataset(
    DATASET_ID,
    DATASET_CONFIG,
    split="train",
    streaming=True,
)

source = source.shuffle(
    seed=SEED,
    buffer_size=2000,
)

print("Датасет:", DATASET_ID)
print("Поднабор:", DATASET_CONFIG)
```

```python
def clean_text(text):
    text = unicodedata.normalize("NFC", text)

    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = text.replace("\u200b", "").replace("\ufeff", "")

    text = "".join(char for char in text if char in "\n\t" or unicodedata.category(char) != "Cc")
    text = re.sub(r"[^\S\n]+", " ", text)
    text = "\n".join(line.strip() for line in text.split("\n"))
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


def suitable_text(text):
    if not 500 <= len(text) <= 30_000:
        return False

    letters = sum(char.isalpha() for char in text)
    russian_letters = len(re.findall(r"[А-Яа-яЁё]", text))

    return letters > 0 and russian_letters / letters >= 0.5


def text_hash(text):
    normalized = re.sub(r"\s+", " ", text).strip()

    return hashlib.sha256(
        normalized.encode("utf-8")
    ).hexdigest()
```

```python
rows = []
seen_hashes = set()
stats = Counter()

with tqdm(total=DOCUMENT_COUNT, desc="Подготовка статей") as bar:
    for item in source:
        stats["прочитано"] += 1

        text = clean_text(item["text"])

        if not suitable_text(text):
            stats["отфильтровано"] += 1
            continue

        digest = text_hash(text)

        if digest in seen_hashes:
            stats["дубликаты"] += 1
            continue

        seen_hashes.add(digest)

        rows.append({
            "id": str(item["id"]),
            "title": item["title"],
            "url": item["url"],
            "text": text,
            "text_hash": digest,
        })

        bar.update(1)

        if len(rows) >= DOCUMENT_COUNT:
            break

print("Документов:", len(rows))
print("Статистика:", dict(stats))

for row in rows[:3]:
    print("\nЗАГОЛОВОК:", row["title"])
    print(row["text"][:700])
    print("-" * 60)
```

```python
documents = Dataset.from_list(rows)

first_split = documents.train_test_split(test_size=0.10,seed=SEED)

second_split = first_split["test"].train_test_split(test_size=0.50,seed=SEED)

corpus = DatasetDict({
    "train": first_split["train"],
    "validation": second_split["train"],
    "test": second_split["test"],
})

hash_sets = {name: set(part["text_hash"]) for name, part in corpus.items()}

for name, part in corpus.items():
    with (DATA_DIR / f"{name}.jsonl").open(
        "w", encoding="utf-8"
    ) as file:
        for row in part:
            file.write(json.dumps(row, ensure_ascii=False) + "\n")

print(corpus)
```

## 2. Обучение токенизатора

```python
def training_text_batches(batch_size=100):
    train = corpus["train"]

    for start in range(0, len(train), batch_size):
        yield train[start:start + batch_size]["text"]


backend = Tokenizer(models.BPE())

backend.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False,use_regex=True)

backend.decoder = decoders.ByteLevel()

bpe_trainer = trainers.BpeTrainer(
    vocab_size=VOCAB_SIZE,
    min_frequency=2,
    show_progress=True,
    special_tokens=[EOS_TOKEN, PAD_TOKEN],
    initial_alphabet=pre_tokenizers.ByteLevel.alphabet(),
)

backend.train_from_iterator(
    training_text_batches(),
    trainer=bpe_trainer,
    length=len(corpus["train"]),
)

tokenizer = PreTrainedTokenizerFast(
    tokenizer_object=backend,
    bos_token=EOS_TOKEN,
    eos_token=EOS_TOKEN,
    pad_token=PAD_TOKEN,
    model_max_length=BLOCK_SIZE,
    clean_up_tokenization_spaces=False,
)

tokenizer.padding_side = "right"
tokenizer.save_pretrained(str(TOKENIZER_DIR))

print("Размер словаря:", len(tokenizer))
print("EOS:", tokenizer.eos_token_id)
print("PAD:", tokenizer.pad_token_id)
```

```python
examples = [
    "Нейросеть обучается на русских текстах.",
    "Ёжик идёт домой.\nНовая строка.",
    "Два  пробела и\tтабуляция.",
    "Python: print('hello') 🙂",
]

for text in examples:
    ids = tokenizer.encode(
        text,
        add_special_tokens=False,
    )

    restored = tokenizer.decode(
        ids,
        skip_special_tokens=False,
        clean_up_tokenization_spaces=False,
    )

    assert restored == text

    print("Текст:", repr(text))
    print("ID:", ids)
    print("Количество токенов:", len(ids))
    print()
```

## 3. Обучение GPT 2

```python
def generate_blocks(documents, tokenizer_json, block_size, eos_id):
    local_backend = Tokenizer.from_str(tokenizer_json)

    buffer = []

    for document in documents:
        ids = local_backend.encode(document["text"], add_special_tokens=False).ids

        ids.append(eos_id)
        buffer.extend(ids)

        full_length = len(buffer) // block_size * block_size

        for start in range(0, full_length, block_size):
            yield {
                "input_ids": buffer[start:start + block_size]
            }

        buffer = buffer[full_length:]


parts = {}

for name, documents_part in corpus.items():
    parts[name] = Dataset.from_generator(
        generate_blocks,
        gen_kwargs={
            "documents": documents_part,
            "tokenizer_json": backend.to_str(),
            "block_size": BLOCK_SIZE,
            "eos_id": tokenizer.eos_token_id,
        },
    )

blocks = DatasetDict(parts)

for name, part in blocks.items():
    print(
        name,
        "— блоков:", len(part),
        "— входных токенов:", len(part) * BLOCK_SIZE,
    )
```

```python
model_parameters = {
    "vocab_size": len(tokenizer),

    "n_positions": BLOCK_SIZE,
    "n_embd": 256,
    "n_layer": 4,
    "n_head": 4,
    "n_inner": 1024,

    "activation_function": "gelu_new",

    "embd_pdrop": 0.1,
    "attn_pdrop": 0.1,
    "resid_pdrop": 0.1,

    "layer_norm_epsilon": 1e-5,
    "initializer_range": 0.02,

    "bos_token_id": tokenizer.bos_token_id,
    "eos_token_id": tokenizer.eos_token_id,
    "pad_token_id": tokenizer.pad_token_id,

    "tie_word_embeddings": True,
    "add_cross_attention": False,
    "use_cache": False,
}

config = GPT2Config(**model_parameters)

print(config)
```

```python
use_bf16 = torch.cuda.is_bf16_supported()
```

```python
set_seed(SEED)

model = GPT2LMHeadModel(config)

model.config.use_cache = False

collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False,
)

sample = collator([
    blocks["train"][i]
    for i in range(min(2, len(blocks["train"])))
])

model.eval()

with torch.no_grad():
    output = model(**sample)

print("Вход:", tuple(sample["input_ids"].shape))
print("Выход:", tuple(output.logits.shape))
print("Проверочный loss:", output.loss.item())
print("Параметров:", sum(p.numel() for p in model.parameters()))

del output, sample
```

```python
updates_per_epoch = math.ceil(
    len(blocks["train"])
    / (MICROBATCH_SIZE * GRADIENT_ACCUMULATION)
)

estimated_updates = updates_per_epoch * EPOCHS

evaluation_interval = min(
    200,
    max(1, updates_per_epoch // 4),
)

warmup_steps = max(1, int(estimated_updates * 0.03))

training_args = TrainingArguments(
    output_dir=str(RUN_DIR),

    num_train_epochs=EPOCHS,

    per_device_train_batch_size=MICROBATCH_SIZE,
    per_device_eval_batch_size=MICROBATCH_SIZE,
    gradient_accumulation_steps=GRADIENT_ACCUMULATION,

    optim="adamw_torch",
    learning_rate=LEARNING_RATE,
    weight_decay=0.01,
    max_grad_norm=1.0,

    warmup_steps=warmup_steps,
    lr_scheduler_type="cosine",

    bf16=use_bf16,
    fp16=False,

    logging_strategy="steps",
    logging_steps=10,

    eval_strategy="steps",
    eval_steps=evaluation_interval,

    save_strategy="no",

    prediction_loss_only=True,

    dataloader_num_workers=0,
    dataloader_pin_memory=True,

    report_to="none",
    push_to_hub=False,

    seed=SEED,
    data_seed=SEED,
)

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=blocks["train"],
    eval_dataset=blocks["validation"],

    data_collator=collator,
    processing_class=tokenizer,
)

print("Устройство:", trainer.args.device)
print(
    "Эффективный пакет:",
    MICROBATCH_SIZE * GRADIENT_ACCUMULATION,
)
print("Оценка каждые", evaluation_interval, "обновлений")
print("BF16:", use_bf16)
```

```python
baseline_metrics = trainer.evaluate()
trainer.save_metrics("baseline", baseline_metrics)

print("Validation до обучения:", baseline_metrics["eval_loss"])

train_result = trainer.train()

trainer.save_metrics("train", train_result.metrics)

validation_metrics = trainer.evaluate()
trainer.save_metrics("validation", validation_metrics)

print("Обучение завершено")
```

```python
inference_tokenizer = tokenizer
inference_model=trainer.model
inference_model.eval()
inference_model.config.use_cache=True

def complete_text(
    prompt,
    max_new_tokens=60,
    temperature=0.8,
    top_p=0.9,
):
    inputs = inference_tokenizer(
        prompt,
        return_tensors="pt",
        return_token_type_ids=False,
    ).to(inference_model.device)
    with torch.inference_mode():
        output = inference_model.generate(
            **inputs,

            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p,

            use_cache=True,

            eos_token_id=inference_tokenizer.eos_token_id,
            pad_token_id=inference_tokenizer.pad_token_id,

            bad_words_ids=[
                [inference_tokenizer.pad_token_id]
            ],
        )

    return inference_tokenizer.decode(
        output[0].tolist(),
        skip_special_tokens=True,
    )
```
